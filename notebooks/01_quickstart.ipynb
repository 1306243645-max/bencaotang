# AI Agent 快速入门
运行你的第一个 Agent 并探索 API。

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from config.settings import settings
from agents.base import BaseAgent
from agents.assistant import AssistantAgent

print(f"Connected to: {settings.ANTHROPIC_BASE_URL}")
print(f"Model: {settings.ANTHROPIC_DEFAULT_MODEL}")
```

```python
# 基础对话
agent = BaseAgent(system="你是一个焊接技术专家，用中文回答。")
resp = agent.chat("焊接机器人和人工焊接相比有哪些优势？")
print(resp.content)
```

```python
# 带记忆的对话助手
assistant = AssistantAgent(name="焊工助手", role="焊接机器人产品顾问")

print(assistant.chat("我们公司焊工现代的焊接机器人出口到澳洲，需要什么认证？").content)
```

```python
# 流式输出
agent = BaseAgent(system="你是一个诗人。")
for chunk in agent.stream("写一首关于机器人和人类合作的四行诗"):
    print(chunk, end="", flush=True)
```
